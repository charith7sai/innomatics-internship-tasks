```python
import re
from collections import Counter

def preprocess_text(text):
    if not text or not isinstance(text, str):
        return ""
    
    # 1. Lowercase & URL/Email removal
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    
    # 2. Remove numbers
    text = re.sub(r'\d+', '', text)
    
    # 3. Handle repeated characters (e.g., goooood -> good)
    # This regex looks for 3+ repetitions and reduces to 2
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # 4. Remove special chars/emojis (keeping basic punctuation if needed)
    text = re.sub(r'[^\w\s]', '', text)
    
    # 5. Tokenize and Filter (Length > 2 or specific exceptions)
    exceptions = {'no', 'not'}
    tokens = [w for w in text.split() if len(w) > 2 or w in exceptions]
    
    return tokens

def full_pipeline(text_list):
    processed_data = {"tokens": [], "clean_sentences": []}
    for sentence in text_list:
        tokens = preprocess_text(sentence)
        processed_data["tokens"].append(tokens)
        processed_data["clean_sentences"].append(" ".join(tokens))
    return processed_data
```
