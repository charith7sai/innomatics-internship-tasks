```python
class SmartLightController:
    def __init__(self, name: str):
        self._name = name
        self._is_on = False

    @property
    def status(self) -> str:
        return "ON" if self._is_on else "OFF"

    def perform_action(self, action: str):
        action = action.upper().strip()
        if action == "ON":
            if self._is_on:
                print(f"Alert: {self._name} is already ON.")
            else:
                self._is_on = True
                self._display_output()
        elif action == "OFF":
            self._is_on = False
            self._display_output()
        else:
            raise ValueError("Invalid Action! Use 'ON' or 'OFF'.")

    def _display_output(self):
        print(f"{self._name} is {self.status}")

# Execution
light = SmartLightController("Bedroom Light")
light.perform_action("ON")
```

```python
class EmployeeIDSystem:
    def __init__(self, name: str, emp_id: str, dept: str):
        self.name = self._validate_string(name)
        self.emp_id = emp_id
        self.dept = self._validate_string(dept)

    def _validate_string(self, value: str):
        if not value.isalpha():
            return value # Simplified for sample, usually raises error
        return value

    def display_id_card(self):
        header = "Employee ID Card"
        print(f"{header}\n{'-'*len(header)}")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.dept}")

# Execution
emp = EmployeeIDSystem("Rahul", "EMP102", "IT")
emp.display_id_card()
```

```python
class MobileContact:
    def __init__(self, name: str, phone: str):
        self.name = name
        self.phone = self._sanitize_phone(phone)

    def _sanitize_phone(self, phone: str) -> str:
        clean_phone = ''.join(filter(str.isdigit, str(phone)))
        if len(clean_phone) != 10:
            raise ValueError("Phone number must be exactly 10 digits.")
        return clean_phone

    def show_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}\nPhone: {self.phone}")

# Execution
contact = MobileContact("Anita", "9876543210")
contact.show_contact()
```

```python
class PriceTagGenerator:
    def __init__(self, p_name: str, price: float):
        self.p_name = p_name
        self.price = float(price)

    def generate_tag(self):
        formatted_price = f"₹{self.price:,.2f}".replace(".00", "")
        print(f"Product: {self.p_name}")
        print(f"Price: {formatted_price}")

# Execution
tag = PriceTagGenerator("Headphones", 2499)
tag.generate_tag()
```

```python
class MovieRatingSystem:
    def __init__(self, title: str, rating: float):
        self.title = title
        self.rating = self._verify_rating(float(rating))

    def _verify_rating(self, val: float) -> float:
        if 0 <= val <= 5:
            return val
        raise ValueError("Rating must be between 0 and 5")

    def __str__(self):
        return f"Movie: {self.title}\nRating: {self.rating} / 5"

# Execution
movie = MovieRatingSystem("Inception", 4.8)
print(movie)
```

```python
class DeliveryManager:
    def __init__(self, customer: str, address: str):
        self.data = {
            "customer": customer,
            "address": address
        }

    def print_shipping_label(self):
        print("Delivery Details")
        print(f"Customer: {self.data['customer']}")
        print(f"Address: {self.data['address']}")

# Execution
delivery = DeliveryManager("Suman", "Hyderabad")
delivery.print_shipping_label()
```
